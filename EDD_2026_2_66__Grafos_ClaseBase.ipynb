```python
class Graph_NDNP:

  def __init__(self):
    self.__listadj = {}

  @property
  def glistadj(self):
    return self.__listadj

  def print(self):

    for vertex in self.__listadj:
      print("Vertex : ", vertex , "- Adyacentes :", self.__listadj[vertex] )

  def add_vertex(self, vertex):
    if vertex not in self.__listadj:
      self.__listadj[vertex] = []
      return True

    return False

  def add_edge(self, vertex1, vertex2):
    if vertex1 == vertex2:
      print("Una arista conecta a 2 vertices diferente")
      return False

    if vertex1 not in self.__listadj or vertex2 not in self.__listadj:
      raise ValueError("Error, los vertices deben estar creados")

    if vertex1 not in self.__listadj[vertex2]:
      self.__listadj[vertex2].append(vertex1)

    if vertex2 not in self.__listadj[vertex1]:
      self.__listadj[vertex1].append(vertex2)


  def delete_edge(self, vertex1, vertex2):
    if vertex1 == vertex2:
      print("Una arista conecta a 2 vertices diferentes")
      return False

    if vertex1 not in self.__listadj or vertex2 not in self.__listadj:
      raise ValueError("Error, los vertices no existen")

    if vertex1 in self.__listadj[vertex2]:
      self.__listadj[vertex2].remove(vertex1)

    if vertex2 in self.__listadj[vertex1]:
      self.__listadj[vertex1].remove(vertex2)


  def delete_vertex(self, vertex):
    if vertex in self.__listadj:
      del self.__listadj[vertex]

      for adjvertex in self.__listadj.values():
        if vertex in adjvertex:
          adjvertex.remove(vertex)


      return True

    raise ValueError("Error, el vertice a eliminar no existe")

```

```python
#Clases Base
from types import new_class
class Node:

  __slots__ = ("__value","__next")

  def __init__(self,value):
    self.__value = value
    self.__next = None

  @property
  def value(self):
    return self.__value

  @property
  def next(self):
    return self.__next


  @value.setter
  def value(self, new_value):
    if new_value is None:
      raise TypeError("El valor No debe ser None/Null.")
    self.__value = new_value

  @next.setter
  def next(self, new_next):
    if new_next is not None and not isinstance(new_next,Node):
      raise TypeError("El next de un nodo, solo puede ser None ó otro Nodo")
    self.__next = new_next


  def __str__(self):
    return str(self.__value)


class slinkedlist:

  __slots__ = ("__head","__tail","__size")

  def __init__(self):
    self.__head = None
    self.__tail = None
    self.__size = 0

  @property
  def head(self):
    return self.__head

  @property
  def tail(self):
    return self.__tail

  @property
  def size(self):
    return self.__size

  @head.setter
  def head(self, new_head):
    if new_head is not None and not isinstance(new_head,Node):
      raise TypeError("La cabeza de una lista enlazada, solo puede ser None ó otro Nodo")
    self.__head = new_head

  @tail.setter
  def tail(self, new_tail):
    if new_tail is not None and not isinstance(new_tail,Node):
      raise TypeError("La cola de una lista enlazada, solo puede ser None ó otro Nodo")
    self.__tail = new_tail

  @size.setter
  def size(self, new_size):
    if not isinstance(new_size,int):
      raise TypeError("El tamaño de una lista enlazada, solo puede ser un numero entero")
    self.__size = new_size

  def __iter__(self):
    cur_node = self.__head
    while cur_node is not None:
      yield cur_node
      cur_node = cur_node.next

  def __str__(self):
     result = [str(node.value) for node in self]
     return ' --> '.join(result)


  def append(self, value):
    new_node = Node(value)

    if self.__tail is None:
      self.__head = new_node
    else:
      self.__tail.next = new_node

    self.__tail = new_node
    self.__size += 1


  def getnodebyindex(self,index):

    if not isinstance(index,int) or (index>self.__size-1) or (index <-1) :
      raise ValueError("Indice por fuera de rango ó no es tipo entero")

    if index == 0:
      return self.__head
    elif index == -1 or index == self.__size-1:
      return self.__tail
    else:
      i_temp = 0
      for cur_node in self:
        if i_temp == index:
          return cur_node
        i_temp += 1


  def pop(self):

    temp_node = self.tail

    if self.__head is None:
      raise TypeError("La lista esta vacía no hay elementos a eliminar")
    elif self.__size == 1:
      self.__head = None
      self.__tail = None
    else:
      prev_tail = self.getnodebyindex(self.__size -2)
      #print("prev_tail",prev_tail)

      prev_tail.next =None
      self.__tail = prev_tail

    self.__size -= 1

    return temp_node.value


class Stack:

  def __init__(self):
    self.__s = slinkedlist()

  def __str__(self):
    result = [str(node.value) for node in self.__s]
    return ' || '.join(result)


  def push(self, e):
    self.__s.append(e)
    return True

  def pop(self):
    if not self.is_empty():
      e = self.__s.pop()
      return e
    else:
      raise ValueError("Error la pila esta vacia")

  def len(self):
    return self.__s.size

  def is_empty(self):
    return self.__s.size == 0

  def top(self):
    if not self.is_empty():
      return self.__s.tail.value
```

```python
def recorrido_bfs(cGraph, vertex):

  auxQueue = Queue()
  visited = set()


  auxQueue.enqueue(vertex)
  visited.add(vertex)

  while not auxQueue.is_empty():

    current_vertex = auxQueue.dequeue()
    print("Recorrido : ", current_vertex)

    for adjvertex in cGraph.glistadj[current_vertex]:
      if adjvertex not in visited:
        auxQueue.enqueue(adjvertex)
        visited.add(adjvertex)


```

```python
def exist_route_dfs(cGraph, source_vertex, target_vertex):

  auxStack = Stack()
  visited = set()


  auxStack.push(source_vertex)
  visited.add(source_vertex)

  while not auxStack.is_empty():

    current_vertex = auxStack.pop()
    print("Recorrido : ", current_vertex)

    for adjvertex in cGraph.glistadj[current_vertex]:
      print("Recorrido vecinos : ", adjvertex)
      if adjvertex not in visited:
        auxStack.push(adjvertex)
        visited.add(adjvertex)

      if adjvertex == target_vertex:
        return True


  return False
```

```python
def count_routes_bfs_3(cGraph, source_vertex, target_vertex): # corregido SI CUENTA TODAS LAS RUTAS Y las retorna

  auxQueue = Queue()
  auxQueue.enqueue((source_vertex,{source_vertex},[source_vertex]))
  count = 0
  result_path = []

  while not auxQueue.is_empty():

    current_vertex, current_visited, current_path = auxQueue.dequeue()
    print("Recorrido/Procesado : ", current_vertex)
    print("Visitados de este recorrido : ", current_visited)
    print("Ruta Actual :", current_path)

    for adjvertex in cGraph.glistadj[current_vertex]:
      print("nodo vecino: ", adjvertex)
      if adjvertex not in current_visited:
        #current_visited.add(adjvertex) # cambia el current_visited de todos, no exclusivo recorrido
        new_visited = current_visited | {adjvertex} # aseguramos q el visitados sea unico por recorrido
        print("nuevos visitados :", new_visited)
        auxQueue.enqueue((adjvertex,new_visited, current_path + [adjvertex] ))


      if adjvertex == target_vertex:
        print("Incrementa contador")
        count +=1
        result_path.append(current_path + [adjvertex])

  return count, result_path
```

```python
count_routes, paths = count_routes_bfs_3(custom_graph5,"X","Y")

print("cuantas rutas existen entre X y Y : ",count_routes)
print("cuales son las rutas que existen entre X y Y : ",paths)
```

```python
def has_cycle(cgraph):

  for vertex in cgraph.glistadj:
    print("Inicia validación para el vertice :", vertex)
    if helper_has_cycle(cgraph, vertex):
      print("Encontro un ciclo en el vertice :", vertex)
      return True
    else:
      print("NO SE Encontro un ciclo en el vertice :", vertex)

  print("El grafo es aciclico")
  return False

def helper_has_cycle(cgraph, source_vertex):

  auxQueue = Queue()
  auxQueue.enqueue((source_vertex,None,{source_vertex}))


  while not auxQueue.is_empty():

    current_vertex, current_parent, current_visited = auxQueue.dequeue()
    print("Recorrido/Procesado : ", current_vertex)
    print("Padre/origen : ", current_parent)
    print("Visitados de este recorrido : ", current_visited)

    for adjvertex in cgraph.glistadj[current_vertex]:
      print("nodo vecino: ", adjvertex)
      if adjvertex not in current_visited:
        #current_visited.add(adjvertex) # cambia el current_visited de todos, no exclusivo recorrido
        new_visited = current_visited | {adjvertex} # aseguramos q el visitados sea unico por recorrido
        print("nuevos visitados :", new_visited)
        auxQueue.enqueue((adjvertex,current_vertex,new_visited))


      if adjvertex == source_vertex and source_vertex != current_parent:
        print("Encontre un ciclo")
        return True

  return False
```

```python
def count_routes_bfs_gp(cGraph, source_vertex, target_vertex): # corregido SI CUENTA TODAS LAS RUTAS Y las retorna + PESO

  auxQueue = Queue()
  auxQueue.enqueue((source_vertex,0,{source_vertex},[source_vertex])) # NUEVO ENCOLAMOS ACUMULADOR DE PESO
  count = 0
  result_path = []

  while not auxQueue.is_empty():

    current_vertex, current_w, current_visited, current_path = auxQueue.dequeue()
    print("Recorrido/Procesado : ", current_vertex)
    print("Peso actual del recorrido :", current_w)
    print("Visitados de este recorrido : ", current_visited)
    print("Ruta Actual :", current_path)

    for adjvertex in cGraph.glistadj[current_vertex]:
      print("nodo vecino: ", adjvertex)
      if adjvertex not in current_visited:
        #current_visited.add(adjvertex) # cambia el current_visited de todos, no exclusivo recorrido
        new_visited = current_visited | {adjvertex} # aseguramos q el visitados sea unico por recorrido
        new_w = current_w + cGraph.glistadj[current_vertex][adjvertex] #LINEA NUEVA PARA PESO
        print("nuevos visitados :", new_visited)
        auxQueue.enqueue((adjvertex,new_w,new_visited, current_path + [adjvertex] ))


      if adjvertex == target_vertex:
        print("Incrementa contador")
        count +=1
        result_path.append((current_path + [adjvertex],new_w))

  return count, result_path
```
